```python
path="federated_101_atlas_atlas.dbo.staged_bw_copa_mach_revrec_base3"
#development_022_silver_finance.lbg.placements_reporting
dest='development_042_silver_sandbox.demand_forecast.sales_bronz_dup'
```

```python
df = spark.table(path)
display(df)
```

```python
import re

# Sanitize column names: replace invalid characters with underscores
df_clean = df
for col_name in df.columns:
    clean_name = re.sub(r'[ ,;{}()\n\t=/]', '_', col_name)
    clean_name = re.sub(r'_+', '_', clean_name).strip('_')
    if clean_name != col_name:
        df_clean = df_clean.withColumnRenamed(col_name, clean_name)

df_clean.write.mode("overwrite").saveAsTable(dest)
```
