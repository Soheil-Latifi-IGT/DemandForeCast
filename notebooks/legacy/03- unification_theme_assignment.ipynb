```python
# Databricks notebook source
# Apply the reviewed theme/mechanic mapping tables to the extracted game records.
# This creates a game-level dataset plus long tables for theme and mechanic analysis.
# It does not modify the extraction source or invent a "goodness" score.

# COMMAND ----------

from pyspark.sql import functions as F
from pyspark.sql.types import (
    ArrayType, IntegerType, StringType, StructField, StructType,
)


# ---------- Configuration ----------

SOURCE_TABLE = "development_042_silver_sandbox.demand_forecast.overview_ctc_line_reel_theme_raw"
TAXONOMY_PREFIX = "development_042_silver_sandbox.demand_forecast.game_taxonomy_unification_v1"

THEME_MAPPING_TABLE = TAXONOMY_PREFIX + "_theme_mapping"
MECHANIC_MAPPING_TABLE = TAXONOMY_PREFIX + "_mechanic_mapping"
THEME_DICTIONARY_TABLE = TAXONOMY_PREFIX + "_theme_dictionary"
MECHANIC_DICTIONARY_TABLE = TAXONOMY_PREFIX + "_mechanic_dictionary"

# New derived outputs. Re-running replaces only these output tables.
OUTPUT_PREFIX = "development_042_silver_sandbox.demand_forecast.canonical_game_dataset_v1"
GAME_DATASET_TABLE = OUTPUT_PREFIX
GAME_THEME_TABLE = OUTPUT_PREFIX + "_themes"
GAME_MECHANIC_TABLE = OUTPUT_PREFIX + "_mechanics"



def write_table(df, table_name):
    df.write.format("delta").mode("overwrite").option(
        "overwriteSchema", "true"
    ).saveAsTable(table_name)
    return spark.table(table_name)


def normalized_text(column):
    return F.lower(F.trim(F.regexp_replace(column.cast("string"), r"\s+", " ")))


def required_columns(df, columns, label):
    missing = set(columns) - set(df.columns)
    if missing:
        raise ValueError(f"{label} is missing columns: {sorted(missing)}")


def apply_overrides(mapping_df, dictionary_table, overrides, taxonomy):
    """Replace AI mappings for selected raw labels with reviewed category choices."""
    if not overrides:
        return mapping_df

    dictionary_df = spark.table(dictionary_table).select(
        "category_id", "category_name", "definition"
    )
    categories = dictionary_df.collect()
    by_name = {}
    for row in categories:
        # Normalize in Python, matching the Spark normalization used for raw labels.
        name_key = " ".join(str(row["category_name"]).strip().lower().split())
        if name_key in by_name:
            raise ValueError(
                f"{taxonomy} dictionary has duplicate canonical name {row['category_name']!r}."
            )
        by_name[name_key] = row

    valid_fields = {row["raw_field"] for row in mapping_df.select("raw_field").distinct().collect()}
    override_rows = []
    seen_override_keys = {}
    for (raw_field, raw_label), target_name in overrides.items():
        if not raw_field or not raw_label or not target_name:
            raise ValueError(f"Empty key/value in {taxonomy} override: {(raw_field, raw_label)!r}")
        if raw_field not in valid_fields:
            raise ValueError(
                f"Unknown {taxonomy} raw_field {raw_field!r}. Available fields: "
                f"{sorted(valid_fields)}"
            )
        normalized_label = " ".join(str(raw_label).strip().lower().split())
        override_key = (str(raw_field), normalized_label)
        if override_key in seen_override_keys:
            raise ValueError(
                f"Conflicting {taxonomy} overrides for {override_key}: "
                f"{seen_override_keys[override_key]!r} and {target_name!r}."
            )
        seen_override_keys[override_key] = target_name
        target_key = " ".join(str(target_name).strip().lower().split())
        if target_key not in by_name:
            raise ValueError(
                f"{taxonomy} override target {target_name!r} is absent from {dictionary_table}. "
                "Add that canonical category to the dictionary first."
            )
        category = by_name[target_key]
        override_rows.append((
            str(raw_field), normalized_label,
            category["category_id"], category["category_name"], category["definition"],
        ))

    override_schema = (
        "raw_field string, normalized_label string, category_id string, "
        "category_name string, definition string"
    )
    override_df = spark.createDataFrame(override_rows, override_schema)
    override_keys = override_df.select("raw_field", "normalized_label").dropDuplicates()

    # Remove all AI assignments for overridden raw labels, including multi-category
    # assignments, then insert exactly the user-selected canonical category.
    untouched = mapping_df.join(
        override_keys, on=["raw_field", "normalized_label"], how="left_anti"
    )
    return untouched.unionByName(override_df, allowMissingColumns=True)


# COMMAND ----------

# ---------- 1. Read source and mapping tables ----------

source = spark.table(SOURCE_TABLE)
required_columns(source, ["game_name", "themes", "mechanics"], "Source table")

theme_map_raw = spark.table(THEME_MAPPING_TABLE)
mechanic_map_raw = spark.table(MECHANIC_MAPPING_TABLE)
mapping_columns = ["raw_field", "normalized_label", "category_id", "category_name", "definition"]
required_columns(theme_map_raw, mapping_columns, "Theme mapping table")
required_columns(mechanic_map_raw, mapping_columns, "Mechanic mapping table")

theme_map = (
    theme_map_raw
    .filter(F.col("raw_field") == "themes")
    .select("raw_field", "normalized_label", "category_id", "category_name", "definition")
    .dropDuplicates(["raw_field", "normalized_label", "category_id"])
)
theme_map = apply_overrides(
    theme_map, THEME_DICTIONARY_TABLE, THEME_OVERRIDES, "theme"
)

mechanic_map = (
    mechanic_map_raw
    .filter(F.col("raw_field").isin(
        "mechanics.source_name", "mechanics.mechanic_family",
        "mechanic_source_names", "mechanic_families"))
    .select("raw_field", "normalized_label", "category_id", "category_name", "definition")
    .dropDuplicates(["raw_field", "normalized_label", "category_id"])
)
mechanic_map = apply_overrides(
    mechanic_map, MECHANIC_DICTIONARY_TABLE, MECHANIC_OVERRIDES, "mechanic"
)

# The extraction table normally stores arrays/structs. These conversions also allow
# the same code to work if the CSV sample was loaded with the nested fields as strings.
if isinstance(source.schema["themes"].dataType, StringType):
    source = source.withColumn("_themes_array", F.from_json("themes", ArrayType(StringType())))
else:
    source = source.withColumn("_themes_array", F.col("themes"))

mechanic_schema = ArrayType(StructType([
    StructField("source_name", StringType(), True),
    StructField("mechanic_family", StringType(), True),
    StructField("mechanic_description", StringType(), True),
    StructField("gameplay_role", StringType(), True),
    StructField("prominence_score", IntegerType(), True),
    StructField("prominence_rationale", StringType(), True),
    StructField("evidence_strength", StringType(), True),
    StructField("evidence_source", StringType(), True),
    StructField("evidence", StringType(), True),
    StructField("variant_notes", StringType(), True),
]))

if isinstance(source.schema["mechanics"].dataType, StringType):
    source = source.withColumn("_mechanics_array", F.from_json("mechanics", mechanic_schema))
else:
    source = source.withColumn("_mechanics_array", F.col("mechanics"))

# A per-row key lets us aggregate labels without accidentally combining different
# source records that happen to share a game name (for example, game variants/files).
source_rows = write_table(
    source.withColumn("_source_row_id", F.monotonically_increasing_id()),
    GAME_DATASET_TABLE + "_temp_source"
)


# COMMAND ----------

# ---------- 2. Apply theme mappings ----------

theme_terms = (
    source_rows
    .select("_source_row_id", "game_name", F.explode_outer("_themes_array").alias("raw_label"))
    .filter(F.col("raw_label").isNotNull() & (F.trim(F.col("raw_label")) != ""))
    .withColumn("normalized_label", normalized_text(F.col("raw_label")))
    .withColumn("raw_field", F.lit("themes"))
)

theme_joined = theme_terms.join(
    theme_map,
    on=["normalized_label", "raw_field"],
    how="left",
)

# Keep one row per source game record and canonical theme, retaining the source labels.
game_theme_categories = (
    theme_joined
    .filter(F.col("category_id").isNotNull())
    .groupBy("_source_row_id", "game_name", "category_id", "category_name", "definition")
    .agg(F.sort_array(F.collect_set("raw_label")).alias("raw_theme_labels"))
)

game_theme_unmapped = (
    theme_joined
    .filter(F.col("category_id").isNull())
    .groupBy("_source_row_id")
    .agg(F.sort_array(F.collect_set("raw_label")).alias("unmapped_theme_labels"))
)

theme_arrays = (
    game_theme_categories
    .groupBy("_source_row_id")
    .agg(F.sort_array(F.collect_list(F.struct(
        "category_id", "category_name", "definition", "raw_theme_labels"
    ))).alias("canonical_themes"))
)


# COMMAND ----------

# ---------- 3. Apply mechanic mappings and retain extracted prominence/evidence ----------

mechanic_records = (
    source_rows
    .select(
        "_source_row_id", "game_name",
        F.posexplode_outer("_mechanics_array").alias("mechanic_index", "mechanic"),
    )
    .filter(F.col("mechanic").isNotNull())
    .select(
        "_source_row_id", "game_name", "mechanic_index",
        F.col("mechanic.source_name").cast("string").alias("source_name"),
        F.col("mechanic.mechanic_family").cast("string").alias("mechanic_family"),
        F.col("mechanic.mechanic_description").cast("string").alias("mechanic_description"),
        F.col("mechanic.gameplay_role").cast("string").alias("gameplay_role"),
        F.col("mechanic.prominence_score").cast("double").alias("prominence_score"),
        F.col("mechanic.prominence_rationale").cast("string").alias("prominence_rationale"),
        F.col("mechanic.evidence_strength").cast("string").alias("evidence_strength"),
        F.col("mechanic.evidence_source").cast("string").alias("evidence_source"),
        F.col("mechanic.evidence").cast("string").alias("evidence"),
        F.col("mechanic.variant_notes").cast("string").alias("variant_notes"),
    )
)

# Match each detailed mechanic record by both its family and its PDF wording.
mechanic_terms = (
    mechanic_records
    .withColumn("_terms", F.array(
        F.struct(F.lit("mechanics.source_name").alias("raw_field"),
                 F.col("source_name").alias("raw_label")),
        F.struct(F.lit("mechanics.mechanic_family").alias("raw_field"),
                 F.col("mechanic_family").alias("raw_label")),
    ))
    .withColumn("_term", F.explode("_terms"))
    .select(
        "_source_row_id", "game_name", "mechanic_index", "source_name", "mechanic_family",
        "mechanic_description", "gameplay_role", "prominence_score", "prominence_rationale",
        "evidence_strength", "evidence_source", "evidence", "variant_notes",
        F.col("_term.raw_field").alias("raw_field"),
        F.col("_term.raw_label").alias("raw_label"),
    )
    .filter(F.col("raw_label").isNotNull() & (F.trim(F.col("raw_label")) != ""))
    .withColumn("normalized_label", normalized_text(F.col("raw_label")))
)

mechanic_joined = mechanic_terms.join(
    mechanic_map,
    on=["raw_field", "normalized_label"],
    how="left",
)

# First collapse family/name matches to one record per game mechanic and canonical category.
# This prevents a record whose name and family both map to the same category from being
# counted twice in the mean prominence score.
mechanic_record_categories = (
    mechanic_joined
    .filter(F.col("category_id").isNotNull())
    .groupBy(
        "_source_row_id", "game_name", "mechanic_index", "category_id",
        "category_name", "definition",
    )
    .agg(
        F.max("prominence_score").alias("record_prominence_score"),
        F.sort_array(F.collect_set("raw_label")).alias("raw_mechanic_labels"),
        F.first("source_name", ignorenulls=True).alias("source_name"),
        F.first("mechanic_family", ignorenulls=True).alias("mechanic_family"),
        F.first("mechanic_description", ignorenulls=True).alias("mechanic_description"),
        F.first("gameplay_role", ignorenulls=True).alias("gameplay_role"),
        F.first("prominence_rationale", ignorenulls=True).alias("prominence_rationale"),
        F.first("evidence_strength", ignorenulls=True).alias("evidence_strength"),
        F.first("evidence_source", ignorenulls=True).alias("evidence_source"),
        F.first("evidence", ignorenulls=True).alias("evidence"),
        F.first("variant_notes", ignorenulls=True).alias("variant_notes"),
    )
)

game_mechanic_categories = (
    mechanic_record_categories
    .groupBy("_source_row_id", "game_name", "category_id", "category_name", "definition")
    .agg(
        F.max("record_prominence_score").alias("prominence_score_max"),
        F.avg("record_prominence_score").alias("prominence_score_mean"),
        F.sort_array(F.collect_set("source_name")).alias("source_names"),
        F.sort_array(F.collect_set("mechanic_family")).alias("extracted_families"),
        F.sort_array(F.collect_set("gameplay_role")).alias("gameplay_roles"),
        F.sort_array(F.array_distinct(F.flatten(F.collect_list("raw_mechanic_labels"))))
            .alias("raw_mechanic_labels"),
        F.collect_list(F.struct(
            "source_name", "mechanic_family", "mechanic_description", "gameplay_role",
            "record_prominence_score", "prominence_rationale", "evidence_strength",
            "evidence_source", "evidence", "variant_notes",
        )).alias("evidence_records"),
    )
)

game_mechanic_unmapped = (
    mechanic_joined
    .filter(F.col("category_id").isNull())
    .groupBy("_source_row_id")
    .agg(F.sort_array(F.collect_set(F.struct("raw_field", "raw_label")))
         .alias("unmapped_mechanic_labels"))
)

mechanic_arrays = (
    game_mechanic_categories
    .groupBy("_source_row_id")
    .agg(F.sort_array(F.collect_list(F.struct(
        "category_id", "category_name", "definition", "prominence_score_max",
        "prominence_score_mean", "source_names", "extracted_families",
        "gameplay_roles", "raw_mechanic_labels", "evidence_records",
    ))).alias("canonical_mechanics"))
)


# COMMAND ----------

# ---------- 4. Write long-form analysis tables and one-row-per-source-record dataset ----------

row_metadata_columns = [c for c in ["game_file", "source_path"] if c in source_rows.columns]
row_metadata = source_rows.select("_source_row_id", *row_metadata_columns)

game_theme_output = game_theme_categories.join(row_metadata, "_source_row_id", "left")
game_mechanic_output = game_mechanic_categories.join(row_metadata, "_source_row_id", "left")

write_table(game_theme_output, GAME_THEME_TABLE)
write_table(game_mechanic_output, GAME_MECHANIC_TABLE)

game_dataset = (
    source_rows
    .join(theme_arrays, "_source_row_id", "left")
    .join(game_theme_unmapped, "_source_row_id", "left")
    .join(mechanic_arrays, "_source_row_id", "left")
    .join(game_mechanic_unmapped, "_source_row_id", "left")
    .drop("_source_row_id", "_themes_array", "_mechanics_array")
    .withColumn("taxonomy_mapping_version", F.lit("v1"))
)

write_table(game_dataset, GAME_DATASET_TABLE)

print("Created:")
print(f"  Game-level dataset: {GAME_DATASET_TABLE}")
print(f"  Game-theme rows:    {GAME_THEME_TABLE}")
print(f"  Game-mechanic rows: {GAME_MECHANIC_TABLE}")
print("The source table was not modified.")

display(spark.table(GAME_DATASET_TABLE).select(
    "game_name", "canonical_themes", "canonical_mechanics",
    "unmapped_theme_labels", "unmapped_mechanic_labels",
    "taxonomy_mapping_version",
))

# ---------- Notes ----------
# prominence_score_max/mean summarize the extraction's 1-5 prominence ratings.
# They indicate how strongly the PDF presents the mechanic; they are not a quality,
# enjoyment, or sales score. Assess mechanic performance separately using outcome data.
# Review the unmapped arrays before using this table for modeling.
```

```python
from pyspark.sql import functions as F

# Existing outputs from your notebooks
BASE_TABLE = (
    "development_042_silver_sandbox.demand_forecast."
    "canonical_game_dataset_v1"
)
THEME_MAPPING_TABLE = (
    "development_042_silver_sandbox.demand_forecast."
    "game_taxonomy_unification_v1_theme_mapping"
)
THEME_DICTIONARY_TABLE = (
    "development_042_silver_sandbox.demand_forecast."
    "game_taxonomy_unification_v1_theme_dictionary"
)
THEME_SCORE_LOOKUP_TABLE = (
    "development_042_silver_sandbox.demand_forecast."
    "game_theme_top40_v1_top40"
)
THEME_GAME_SCORES_TABLE = (
    "development_042_silver_sandbox.demand_forecast."
    "game_theme_top40_v1_game_scores"
)
BRONZ_TABLE = (
    "development_042_silver_sandbox.demand_forecast.game_attr_ai_bronz"
)

# Older theme names that were explicitly renamed in your reviewed overrides.
THEME_SCORE_NAME_OVERRIDES = {
    "Detective and Mystery": "Mystery and Detective",
    "Diamonds and Jewels": "Gems and Jewels",
    "Historical Settings": "Ancient and Historical",
}

# Used to populate the derived persistence score when one of these
# canonical mechanic categories exists in the unified mechanic results.
PERSISTENCE_CATEGORY_NAMES = {
    "perceived persistence",
    "persistence mechanics",
}

def norm(value):
    return " ".join(str(value).strip().lower().split())

base = spark.table(BASE_TABLE)

theme_scores = spark.table(THEME_GAME_SCORES_TABLE)
theme_lookup = spark.table(THEME_SCORE_LOOKUP_TABLE)
theme_mapping = (
    spark.table(THEME_MAPPING_TABLE)
    .filter(F.col("raw_field") == "themes")
    .select("normalized_label", "category_name")
    .dropDuplicates()
)

# Match old scored theme names to the current canonical theme categories.
mapping_by_label = {}
for row in theme_mapping.collect():
    mapping_by_label.setdefault(norm(row["normalized_label"]), set()).add(
        row["category_name"]
    )

canonical_names = {
    norm(row["category_name"]): row["category_name"]
    for row in spark.table(THEME_DICTIONARY_TABLE)
        .select("category_name").distinct().collect()
}

score_parts = []
unmatched_theme_names = []

for row in theme_lookup.select("theme_name", "score_column").collect():
    old_name = row["theme_name"]
    score_column = row["score_column"]

    if score_column not in theme_scores.columns:
        continue

    candidates = mapping_by_label.get(norm(old_name), set())

    if not candidates and norm(old_name) in canonical_names:
        candidates = {canonical_names[norm(old_name)]}

    if not candidates:
        renamed = THEME_SCORE_NAME_OVERRIDES.get(old_name)
        if renamed and norm(renamed) in canonical_names:
            candidates = {canonical_names[norm(renamed)]}

    if len(candidates) != 1:
        unmatched_theme_names.append(old_name)
        continue

    score_parts.append(
        theme_scores.select(
            F.col("game_key").alias("_game_key"),
            F.lit(next(iter(candidates))).alias("canonical_theme"),
            F.col(score_column).cast("double").alias("theme_score"),
        )
    )

if not score_parts:
    raise ValueError(
        "No old theme scores matched the current canonical theme names. "
        "Review the theme-name mapping before creating bronz."
    )

theme_score_long = score_parts[0]
for part in score_parts[1:]:
    theme_score_long = theme_score_long.unionByName(part)

# If multiple old theme labels were unified, keep the strongest supported score.
canonical_theme_scores = (
    theme_score_long
    .groupBy("_game_key", "canonical_theme")
    .agg(F.max("theme_score").alias("theme_score"))
    .groupBy("_game_key")
    .agg(
        F.sort_array(
            F.collect_list(F.struct("canonical_theme", "theme_score"))
        ).alias("canonical_theme_scores")
    )
)

theme_score_metadata = (
    theme_scores
    .select(
        F.col("game_key").alias("_game_key"),
        "theme_scoring_status",
        "theme_expected_pages",
        "theme_scored_pages",
        "theme_taxonomy_id",
    )
    .dropDuplicates(["_game_key"])
)

# Join scores to each source game row. The original fields in BASE_TABLE
# (including CTC, ways, lines, reels, and game attributes) remain in the output.
base_keyed = base.withColumn(
    "_game_key",
    F.lower(F.trim(F.regexp_replace(F.col("game_name"), r"\s+", " ")))
)

bronz = (
    base_keyed
    .join(canonical_theme_scores, on="_game_key", how="left")
    .join(theme_score_metadata, on="_game_key", how="left")
)

bronz = bronz.withColumn(
    "mechanic_scores",
    F.transform(
        "canonical_mechanics",
        lambda m: F.struct(
            m.category_id.alias("category_id"),
            m.category_name.alias("mechanic"),
            m.prominence_score_mean.alias("score_mean"),
            m.prominence_score_max.alias("score_max"),
        ),
    ),
)

persistence_names_sql = ", ".join(
    "'" + name.replace("'", "''") + "'"
    for name in sorted(PERSISTENCE_CATEGORY_NAMES)
)

bronz = bronz.withColumn(
    "perceived_persistence_score",
    F.expr(f"""
        aggregate(
            filter(
                canonical_mechanics,
                m -> lower(trim(m.category_name)) IN ({persistence_names_sql})
            ),
            CAST(NULL AS DOUBLE),
            (current_score, m) ->
                CASE
                    WHEN m.prominence_score_mean IS NULL THEN current_score
                    WHEN current_score IS NULL THEN m.prominence_score_mean
                    ELSE greatest(current_score, m.prominence_score_mean)
                END
        )
    """),
).drop("_game_key")

(
    bronz.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(BRONZ_TABLE)
)

display(spark.table(BRONZ_TABLE))
```
